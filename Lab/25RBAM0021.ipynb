4. Write a Python program to check whether a given number is a prime number.

 Algorithm :
step 1:Start           
step 2:Input a number from the user.                      
step 3:Check special cases:                       
        If the number is less than or equal to 1 → it’s not prime.                     
step 4:Loop through possible divisors                             
        For each integer i from 2 to √n (square root of the number):                         
        If n % i == 0 → the number is not prime.                        
step 5:If no divisors found → the number is prime.                               
step 6:Display the result.
step 6:End

```python
num = int(input("Enter a number: "))

if num <= 1:
    print(num, "is not a prime number.")
else:
    
    is_prime = True
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    
    if is_prime:
        print(num, "is a prime number.")
    else:
        print(num, "is not a prime number.")
```

6. Write a Python program to check whether a number is a palindrome.

```python

num = int(input("Enter a number: "))

original = num
rev = 0

while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10

if original == rev:
    print(original, "is a palindrome number.")
else:
    print(original, "is not a palindrome number.")
```

7. Write a Python program to check whether a given string is a palindrome

Algorithm :
step 1:Start
step 2:Input a string from the user.
step 3:Reverse the string.
        -You can reverse using slicing (string[::-1]).
step 4:Compare the original string with the reversed string.
        -If they are equal → the string is a palindrome.
        -Else → the string is not a palindrome.
step 5:Display the result.
step 6:End

```python

text = input("Enter a string: ")

rev_text = text[::-1]

if text == rev_text:
    print(text, "is a palindrome string.")
else:
    print(text, "is not a palindrome string.")
```

8. Write a Python program to print the Fibonacci series up to N terms.

```python

N = int(input("Enter the number of terms: "))

a, b = 0, 1

if N <= 0:
    print("Please enter a positive integer.")
elif N == 1:
    print("Fibonacci series up to 1 term:")
    print(a)
else:
    print("Fibonacci series up to", N, "terms:")
    print(a, b, end=" ")
    for i in range(2, N):
        next_term = a + b
        print(next_term, end=" ")
        a, b = b, next_term
```

9. Write a Python program to find the sum of digits of a number

```python

num = int(input("Enter a number: "))

sum_of_digits = 0
temp = num

while temp > 0:
    digit = temp % 10
    sum_of_digits += digit
    temp = temp // 10
print("The sum of digits of", num, "is:", sum_of_digits)
```

22. Write a Python program to create a Library Book Management System using functions.

```python
ibrary = []  

def add_book(title, author):
    book = {"title": title, "author": author, "available": True}
    library.append(book)
    print(f"✅ Book '{title}' by {author} added successfully!")

def display_books():
    if not library:
        print("📭 No books in the library.")
    else:
        print("\n📚 Library Books:")
        for idx, book in enumerate(library, start=1):
            status = "Available" if book["available"] else "Borrowed"
            print(f"{idx}. {book['title']} by {book['author']} - {status}")

def search_book(title):
    for book in library:
        if book["title"].lower() == title.lower():
            status = "Available" if book["available"] else "Borrowed"
            print(f"🔍 Found: {book['title']} by {book['author']} - {status}")
            return
    print("❌ Book not found.")

def borrow_book(title):
    for book in library:
        if book["title"].lower() == title.lower():
            if book["available"]:
                book["available"] = False
                print(f"📖 You borrowed '{book['title']}'.")
            else:
                print("⚠️ Book is already borrowed.")
            return
    print("❌ Book not found.")

def return_book(title):
    for book in library:
        if book["title"].lower() == title.lower():
            if not book["available"]:
                book["available"] = True
                print(f"📦 You returned '{book['title']}'.")
            else:
                print("⚠️ Book was not borrowed.")
            return
    print("❌ Book not found.")

while True:
    print("\n===== Library Menu =====")
    print("1. Add Book")
    print("2. Display Books")
    print("3. Search Book")
    print("4. Borrow Book")
    print("5. Return Book")
    print("6. Exit")

    choice = input("Enter your choice (1-6): ")

    if choice == "1":
        title = input("Enter book title: ")
        author = input("Enter author name: ")
        add_book(title, author)
    elif choice == "2":
        display_books()
    elif choice == "3":
        title = input("Enter book title to search: ")
        search_book(title)
    elif choice == "4":
        title = input("Enter book title to borrow: ")
        borrow_book(title)
    elif choice == "5":
        title = input("Enter book title to return: ")
        return_book(title)
    elif choice == "6":
        print("👋 Exiting Library System. Goodbye!")
        break
    else:
        print("❌ Invalid choice. Please try again.")
```

25. Write a Python program using modular programming principles and demonstrate:
•	Input validation
•	Testing (minimum 3 test cases)
•	Debugging practice with comments

```python
# calculator_module.py - Demonstrating modular programming, input validation, testing, debugging

# ------------------ Operations ------------------
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Error! Division by zero."
    return a / b

# ------------------ Input Validation ------------------
def get_number(prompt):
    while True:
        try:
            return float(input(prompt))   # Accepts integers and decimals
        except ValueError:
            print("❌ Invalid input! Please enter a number.")

# ------------------ Menu ------------------
def display_menu():
    print("\n===== Calculator Menu =====")
    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Exit")

# ------------------ Main Program ------------------
def main():
    while True:
        display_menu()
        choice = input("Enter your choice (1-5): ")

        if choice == "5":
            print("👋 Exiting program. Goodbye!")
            break

        # Debugging checkpoint: print choice
        print(f"[DEBUG] User selected option: {choice}")

        num1 = get_number("Enter first number: ")
        num2 = get_number("Enter second number: ")

        # Debugging checkpoint: print inputs
        print(f"[DEBUG] Numbers entered: {num1}, {num2}")

        if choice == "1":
            print(f"✅ Result: {add(num1, num2)}")
        elif choice == "2":
            print(f"✅ Result: {subtract(num1, num2)}")
        elif choice == "3":
            print(f"✅ Result: {multiply(num1, num2)}")
        elif choice == "4":
            print(f"✅ Result: {divide(num1, num2)}")
        else:
            print("❌ Invalid choice. Try again.")

# ------------------ Testing ------------------
def test_cases():
    print("\n===== Running Test Cases =====")
    # Test Case 1: Addition
    assert add(10, 5) == 15, "[TEST FAILED] Addition"
    print("[TEST PASSED] Addition")

    # Test Case 2: Division by non-zero
    assert divide(20, 4) == 5, "[TEST FAILED] Division"
    print("[TEST PASSED] Division")

    # Test Case 3: Division by zero
    assert divide(10, 0) == "Error! Division by zero.", "[TEST FAILED] Division by zero"
    print("[TEST PASSED] Division by zero")

# ------------------ Run Program ------------------
if __name__ == "__main__":
    test_cases()   # Run tests first
    main()         # Then start interactive program 
```

26. Write a Python project for a User Registration System with input validation, testing, and debugging documentation.

```python
import re   # For email validation

# ------------------ Data Store ------------------
users = []  # List to store registered users

# ------------------ Validation Functions ------------------
def validate_username(username):
    # Debugging checkpoint
    print(f"[DEBUG] Validating username: {username}")
    if not username:
        return "❌ Username cannot be empty."
    for user in users:
        if user["username"].lower() == username.lower():
            return "❌ Username already exists."
    return "✅ Valid"

def validate_password(password):
    # Debugging checkpoint
    print(f"[DEBUG] Validating password: {password}")
    if len(password) < 6:
        return "❌ Password must be at least 6 characters."
    if not re.search(r"[A-Za-z]", password) or not re.search(r"[0-9]", password):
        return "❌ Password must contain both letters and numbers."
    return "✅ Valid"

def validate_email(email):
    # Debugging checkpoint
    print(f"[DEBUG] Validating email: {email}")
    pattern = r'^[\w\.-]+@[\w\.-]+\.\w+$'
    if not re.match(pattern, email):
        return "❌ Invalid email format."
    return "✅ Valid"

# ------------------ Registration Function ------------------
def register_user(username, password, email):
    u_check = validate_username(username)
    p_check = validate_password(password)
    e_check = validate_email(email)

    if u_check == "✅ Valid" and p_check == "✅ Valid" and e_check == "✅ Valid":
        users.append({"username": username, "password": password, "email": email})
        print(f"🎉 User '{username}' registered successfully!")
        return True
    else:
        print(u_check)
        print(p_check)
        print(e_check)
        return False

# ------------------ Display Users ------------------
def display_users():
    print("\n===== Registered Users =====")
    if not users:
        print("📭 No users registered yet.")
    else:
        for idx, user in enumerate(users, start=1):
            print(f"{idx}. {user['username']} - {user['email']}")

# ------------------ Testing ------------------
def test_cases():
    print("\n===== Running Test Cases =====")

    # Test Case 1: Valid registration
    assert register_user("Sam123", "pass123", "sam@example.com") == True, "[TEST FAILED] Valid registration"
    print("[TEST PASSED] Valid registration")

    # Test Case 2: Duplicate username
    assert register_user("Sam123", "newpass1", "sam2@example.com") == False, "[TEST FAILED] Duplicate username"
    print("[TEST PASSED] Duplicate username")

    # Test Case 3: Invalid password
    assert register_user("Alex", "123", "alex@example.com") == False, "[TEST FAILED] Invalid password"
    print("[TEST PASSED] Invalid password")

# ------------------ Main Program ------------------
def main():
    test_cases()  # Run tests first

    while True:
        print("\n===== User Registration Menu =====")
        print("1. Register User")
        print("2. Display Users")
        print("3. Exit")

        choice = input("Enter your choice (1-3): ")

        if choice == "1":
            username = input("Enter username: ")
            password = input("Enter password: ")
            email = input("Enter email: ")
            register_user(username, password, email)
        elif choice == "2":
            display_users()
        elif choice == "3":
            print("👋 Exiting system. Goodbye!")
            break
        else:
            print("❌ Invalid choice. Try again.")

if __name__ == "__main__":
    main()
```
